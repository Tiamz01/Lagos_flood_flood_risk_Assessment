```python
import geopandas as gpd
import matplotlib.pyplot as plt

lgas_path = BOUNDARIES_DIR / "lagos_lgas.geojson"
if lgas_path.exists():
    gdf_lgas = gpd.read_file(lgas_path)
    fig, ax = plt.subplots(figsize=(12, 6))
    gdf_lgas.plot(ax=ax, edgecolor="black", facecolor="lightblue", alpha=0.6)
    gdf_lgas.boundary.plot(ax=ax, color="navy", linewidth=1)
    plt.title("Lagos State Local Government Areas (LGAs) - Baseline Boundary Map", fontsize=14)
    plt.xlabel("Longitude")
    plt.ylabel("Latitude")
    plt.grid(True, linestyle="--", alpha=0.5)
    plt.show()
else:
    print("Boundary file not yet generated. Run get_lagos_boundaries() first.")
```
